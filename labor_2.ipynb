**Problem 2**

**Part 1 - We consider two infections:**

- $I$: A patient has influenza.
- $S$: A patient has strep throat.

$$
P(I)=0.08=8\%
$$

$$
P(S)=0.05=5\%
$$

The two infections are assumed to be **statistically independent**.

For independent events => the probability that both occur is:

$$
P(A\cap B)=P(A)\cdot P(B)
$$

**a) Probability of Having Both Infections**

We want the probability that a patient has both influenza and strep throat:

$$
P(I\cap S)
$$

Both infections are independent => we multiply their probabilities:

$$
P(I\cap S)=P(I)\cdot P(S)=0.08\cdot0.05=0.004=0.4\%
$$

**Explanation:**

Because the infections are independent, having influenza does not change the probability of having strep throat => we multiply their probabilities.

---

**b) Probability of Having at Least One Infection**

This includes influenza only, strep throat only, or both infections.

The probability corresponds to the union:

$$
P(I\cup S)
$$

We apply the general addition rule:

$$
P(I\cup S)=P(I)+P(S)-P(I\cap S)
$$

$$
P(I\cup S)=0.08+0.05-0.004=0.126=12.6\%
$$

**Explanation:**

We subtract the intersection because patients with both infections are counted twice when adding $P(I)$ and $P(S)$.

**Alternative with the complement rule:**

We calculate the probability of having neither infection and subtract it from $1$.

$$
P(\neg I\cap\neg S)=P(\neg I)\cdot P(\neg S)
$$

The complement probabilities are:

$$
P(\neg I)=1-0.08=0.92
$$

$$
P(\neg S)=1-0.05=0.95
$$

Substituting:

$$
P(\neg I\cap\neg S)=0.92\cdot0.95=0.874
$$

Therefore:

$$
P(I\cup S)=1-0.874=0.126=12.6\%
$$

---

**c) Probability of Having Exactly One Infection**

We want influenza without strep throat, or strep throat without influenza:

$$
P(\text{exactly one})=P(I\cap\neg S)+P(S\cap\neg I)
$$

**Step 1: Calculate influenza without strep throat.**

The events are independent =>

$$
P(I\cap\neg S)=P(I)\cdot P(\neg S)=0.08\cdot0.95=0.076=7.6\%
$$

**Step 2: Calculate strep throat without influenza.**

$$
P(S\cap\neg I)=P(S)\cdot P(\neg I)=0.05\cdot0.92=0.046=4.6\%
$$

**Step 3: Add the probabilities.**

$$
P(\text{exactly one})=0.076+0.046=0.122=12.2\%
$$

**Alternative formula:**

The probability of exactly one event can also be calculated using:

$$
P(\text{exactly one})=P(I)+P(S)-2P(I\cap S)
$$

$$
=0.08+0.05-2(0.004)=0.122=12.2\%
$$

**Explanation:**

When we add $P(I)+P(S)$, patients with both infections are counted twice.

But we want only patients with exactly one infection, so those with both infections must be excluded completely => we subtract the intersection twice.

**Part 2 - Rapid Influenza Test**

The rapid influenza test has:

- **Sensitivity:** $70\%$
- **Specificity:** $95\%$

**Sensitivity** = the probability that the test is positive when the patient actually has influenza - true positive.

$$
\text{Sensitivity}=P(+\mid I)=0.70
$$

The remaining $30\%$ are false negatives:

$$
P(-\mid I)=1-0.70=0.30
$$

**Specificity** = the probability that the test is negative when the patient does not have influenza - true negative.

$$
\text{Specificity}=P(-\mid\neg I)=0.95
$$

The remaining $5\%$ receive false positive results:

$$
P(+\mid\neg I)=1-0.95=0.05
$$

We want to calculate the probability that a patient actually has influenza **after receiving a positive test result**:

$$
P(I\mid +)
$$

![Screenshot 2026-11-10 080442.png](pictures/Screenshot%202026-11-10%20080442.png)

**Bayes Theorem is used to find the probability of an event given that another event has already occurred.**
So we use **Bayes' Theorem**:

$$
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}
$$

Applied to influenza:

$$
P(I\mid +)=\frac{P(+\mid I)P(I)}{P(+)}
$$

**A positive test may occur in two situations:**

1. true pos
2. false pos

**Therefore, we use the Law of Total Probability:**

$$
P(+)=P(+\mid I)P(I)+P(+\mid\neg I)P(\neg I)
$$

**And we combine this with Bayes' Theorem:**

$$
P(I\mid +)=
\frac{P(+\mid I)P(I)}
{P(+\mid I)P(I)+P(+\mid\neg I)P(\neg I)}
$$

This formula will be used for both winter and summer.

---

**d) Probability of Having Influenza After a Positive Test in Winter**

During winter, the prevalence of influenza is:

$$
P(I)=0.08
$$

So the probability of not having influenza is:

$$
P(\neg I)=1-0.08=0.92
$$

The test characteristics are:

$$
P(+\mid I)=0.70
$$

$$
P(+\mid\neg I)=0.05
$$

We apply **Bayes' Theorem**:

$$
P(I\mid +)=
\frac{P(+\mid I)P(I)}
{P(+\mid I)P(I)+P(+\mid\neg I)P(\neg I)}
$$

Substituting the values:

$$
P(I\mid +)=
\frac{0.70\cdot0.08}
{0.70\cdot0.08+0.05\cdot0.92}
=
\frac{0.056}{0.102}
\approx0.5490=54.90\%
$$

**Explanation:**

- $0.056$ = true positive.
- $0.046$ = false positive.
- $0.102$ = total probability of receiving a positive test result.

We calculate the total probability of a positive test:

$$
P(+)=0.056+0.046=0.102
$$

Then divide the true positive probability by the total positive probability:

$$
P(I\mid +)=\frac{0.056}{0.102}\approx54.90\%
$$

**Interpretation:**

A patient who tests positive during winter has approximately a **54.9% probability of actually having influenza**.
A positive result does not guarantee infection because false positives also occur.

---

**e) Probability of Having Influenza After a Positive Test in Summer**

During summer, influenza is much less common:

$$
P(I)=0.005=0.5\%
$$

So the probability of not having influenza is:

$$
P(\neg I)=1-0.005=0.995
$$

The sensitivity and specificity remain the same:

We use **Bayes' Theorem** again, but with the summer prevalence:

$$
P(I\mid +)=
\frac{P(+\mid I)P(I)}
{P(+\mid I)P(I)+P(+\mid\neg I)P(\neg I)}
$$

Substituting the values:

$$
P(I\mid +)=
\frac{0.70\cdot0.005}
{0.70\cdot0.005+0.05\cdot0.995}
=
\frac{0.0035}{0.05325}
\approx0.0657=6.57\%
$$

**Explanation:**

- $0.0035$ = true positive.
- $0.04975$ = false positive.
- $0.05325$ = total probability of a positive test.

We divide the probability of a true positive by the probability of all positive results to determine how likely a positive patient is to actually have influenza.

A patient who tests positive during summer has only a **6.57% probability of actually having influenza**.
Even though the test has $95\%$ specificity, most positive results in summer are false positives because influenza is very rare.

---

**f) Is the CDSS Alert "Probable Influenza" Sensible?**

Not in every case. The CDSS should consider disease prevalence, not only the positive test result.

**In winter:**

$$
P(I\mid +)\approx54.90\%
$$

**In summer:**

$$
P(I\mid +)\approx6.57\%
$$

Although the test has the same sensitivity and specificity in both seasons, the probability of actually having influenza is very different.

**Explanation:**

Bayes' Theorem takes into account the **prior probability** of the disease.
In winter, influenza is more common => a positive result is more likely to be a true positive.
In summer, influenza is rare => false positives make up a much larger proportion of positive results.

**Conclusion:**

The alert **"Probable influenza" is not always sensible**, especially during summer when the probability of actually having influenza is only $6.57\%$.
The CDSS should consider seasonal prevalence and patient-specific clinical information before issuing the alert.
Also, the independence assumption from Part 1 may not always hold in reality because the two infections can be influenced by shared risk factors.


```python
# part 1

P_I = 0.08
P_S = 0.05

# a.
P_both = P_I * P_S
print(f"(a) Both infections: {P_both:.4f} = {P_both:.2%}")

# b.
P_at_least_one = P_I + P_S - P_both
print(f"(b) At least one infection: {P_at_least_one:.4f} = {P_at_least_one:.2%}")

# alternative
P_neither = (1 - P_I) * (1 - P_S)
print(f"    Complement method: {1 - P_neither:.4f}")

# c.
P_only_I = P_I * (1 - P_S)
P_only_S = P_S * (1 - P_I)
P_exactly_one = P_only_I + P_only_S
print(f"(c) Exactly one infection: {P_exactly_one:.4f} = {P_exactly_one:.2%}")


# part

sensitivity = 0.70
specificity = 0.95
false_positive_rate = 1 - specificity

# Bayes' Theorem: P(I | positive)
def probability_influenza_given_positive(prevalence):
    true_positive = sensitivity * prevalence
    false_positive = false_positive_rate * (1 - prevalence)
    return true_positive / (true_positive + false_positive)


# d. prob after a + test in winter
P_winter = probability_influenza_given_positive(0.08)
print(f"\n(d) Positive test in winter: {P_winter:.4f} = {P_winter:.2%}")

# e. prob after a + test in summer
P_summer = probability_influenza_given_positive(0.005)
print(f"(e) Positive test in summer: {P_summer:.4f} = {P_summer:.2%}")
```

![Screenshot 2026-11-10 083105.png](pictures/Screenshot%202026-11-10%20083105.png)
We are given three sets:

- $C$ = CLINICAL-FINDINGS (symptoms observed in the patient).
- $B$ = BEHAVIORAL-SYMPTOMS = {nausea, dizziness, loss of consciousness}.
- $N$ = NEURAL-SYMPTOMS = {headache, head wound, memory loss}.

A patient has a concussion if **at least one behavioral symptom AND at least two neural symptoms** are present.

**1. Behavioral condition:**

$$
|C\cap B|\geq1
$$

**2. Neural condition:**

$$
|C\cap N|\geq2
$$

**Final formula:**

$$
\boxed{
\operatorname{Concussion}(C)
\iff
(|C\cap B|\geq1)\land(|C\cap N|\geq2)
}
$$

**Explanation:**

- $\cap$ => symptoms present in both sets.
- $|\cdot|$ => number of symptoms in the intersection.
- $\geq$ = at least.
- $\land$ = AND => both conditions must be satisfied.
- $\iff$ = if and only if.

Therefore, the patient satisfies the concussion rule **only when both conditions are true**.


```python
behavioral = {"nausea", "dizziness", "loss of consciousness"}
neural = {"headache", "head wound", "memory loss"}

def has_concussion(findings):
    return len(findings & behavioral) >= 1 and len(findings & neural) >= 2

# ex patients
patient1 = {"nausea", "headache", "memory loss"}
patient2 = {"dizziness", "headache"}
patient3 = {"headache", "head wound", "memory loss"}

print("Patient 1:", has_concussion(patient1))
print("Patient 2:", has_concussion(patient2))
print("Patient 3:", has_concussion(patient3))
```


**Problem 4 - Greedy Algorithm (Set Cover)**
![Screenshot 2026-11-10 083203.png](pictures/Screenshot%202026-11-10%20083203.png)
We are given six pathogens and six antibiotics. We need to cover all pathogens using the minimum number of antibiotics.

**1. Greedy Algorithm**

The greedy algorithm starts with an empty solution set:

$$
S_0=\varnothing
$$

At each iteration => select the antibiotic covering the **largest number of currently uncovered pathogens**.
The algorithm stops when all pathogens are covered.

**Greedy Iterations:**

| Iteration | Selected antibiotic | Newly covered | Total covered | Solution set |
|---|---|---|---|---|
| 0 | None | $\varnothing$ | $\varnothing$ | $\varnothing$ |
| 1 | C-Cillin | $\{A,C,F\}$ | $\{A,C,F\}$ | $\{C\}$ |
| 2 | A-Cillin | $\{B\}$ | $\{A,B,C,F\}$ | $\{C,A\}$ |
| 3 | B-Cillin | $\{E\}$ | $\{A,B,C,E,F\}$ | $\{C,A,B\}$ |
| 4 | D-Cillin | $\{D\}$ | $\{A,B,C,D,E,F\}$ | $\{C,A,B,D\}$ |

Extra: when multiple antibiotics cover the same number of uncovered pathogens, ties can be broken arbitrarily.

**Greedy result:**

$$
|S_{\text{greedy}}|=4
$$

All six pathogens are covered, but this solution is **not optimal**.

---

**2. Optimal Solution**

We can cover all pathogens using only three antibiotics:

| Iteration | Selected antibiotic | Newly covered | Total covered |
|---|---|---|---|
| 0 | None | $\varnothing$ | $\varnothing$ |
| 1 | A-Cillin | $\{A,B\}$ | $\{A,B\}$ |
| 2 | B-Cillin | $\{E,F\}$ | $\{A,B,E,F\}$ |
| 3 | D-Cillin | $\{C,D\}$ | $\{A,B,C,D,E,F\}$ |

$$
|S_{\text{optimal}}|=3
$$

So this will be the optimal solution, because it has the **minimum number of antibiotics** that cover all pathogens.

**The greedy algorithm makes the locally best choice at each iteration without considering future selections.**
C-Cillin initially covers the most pathogens. But after selecting it, there are still three pathogens left uncovered.
No remaining antibiotic covers more than one of them => three additional antibiotics are required.
However, the optimal combination covers all pathogens using only three antibiotics, so the greedy solution is valid but not optimal.

```python
pathogens = {"A", "B", "C", "D", "E", "F"}

antibiotics = {
    "A-Cillin": {"A", "B"},
    "B-Cillin": {"E", "F"},
    "C-Cillin": {"A", "C", "F"},
    "D-Cillin": {"C", "D"},
    "E-Cillin": {"B"},
    "F-Cillin": {"D"}
}

covered = set()
selected = []

while covered != pathogens:
    best = max(
        antibiotics,
        key=lambda name: len(antibiotics[name] - covered)
    )

    newly_covered = antibiotics[best] - covered
    covered.update(newly_covered)
    selected.append(best)

    print(f"{best}: new = {sorted(newly_covered)}, "
          f"covered = {sorted(covered)}")

    del antibiotics[best]

print("\nGreedy solution:", selected)
print("Total antibiotics:", len(selected))
```
