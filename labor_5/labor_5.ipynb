## **Problem 1**

a. In the Laplace mechanism, $X$ is drawn from a Laplace distribution centered at 0, with a scale parameter of $\frac{\Delta f}{\epsilon}$, where $\Delta f$ is the sensitivity of the function and $\epsilon$ is the privacy parameter.

b. In the Geometric mechanism, $X$ is drawn from a two-sided geometric distribution. This can be seen as the discrete version of the Laplace distribution, since it is used to generate integer-valued noise.

c. According to the sequential composition property of differential privacy, the privacy parameters of the two mechanisms are added together. Therefore, the resulting privacy parameter is $\epsilon = 2 + 0.1 = 2.1$.

d. Parametric synthetic data generation means generating new data based on an assumed statistical model. The parameters of the model are estimated from the original data, and new data is then generated using the fitted model.

e. Non-parametric synthetic data generation means generating new data without assuming a specific parametric distribution. Instead, the patterns and relationships are learned directly from the original data and used to generate new data.

f. The odds ratio distance is the absolute difference between the odds ratios obtained from the original and the noisy data. For example, if the odds ratio is 1.56 on the original data and 0.54 on the noisy data, the odds ratio distance is $|1.56 - 0.54| = 1.02$.

g. The p-score indicates whether the statistical significance of a result changes between the original and the noisy data. It is 1 if a result changes from significant to not significant or vice versa, and 0 otherwise. In the experiment presented in the lecture, these values are summed over all 25 runs.

**c. Re-identification risk analysis**

To compare the re-identification risk of the original and transformed datasets, we used the Analyze risk section of the ARX tool.

According to the ARX documentation, there are three main attacker models:

- Prosecutor: the attacker knows that a specific person is included in the dataset and tries to identify their record.
- Journalist: the attacker tries to identify a person without knowing whether that person is included in the dataset.
- Marketer: the attacker tries to identify as many individuals as possible from the dataset.

After applying k-anonymity with $k = 5$, we obtained the following results:

| Risk measure | Original data | Transformed data |
|---|---:|---:|
| Lowest prosecutor risk | 2.22222% | 0.11792% |
| Average prosecutor risk | 64.65752% | 2.17927% |
| Highest prosecutor risk | 100% | 20% |
| Records affected by lowest risk | 0.14919% | 3.07491% |
| Records affected by highest risk | 51.42895% | 1.14221% |
| Estimated prosecutor risk | 100% | 20% |
| Estimated journalist risk | 100% | 20% |
| Estimated marketer risk | 64.65752% | 2.17927% |
| Sample uniques | 51.42895% | 0% |
| Estimated population uniques | 3.88962% | 0% |

We can observe that the re-identification risk decreases considerably after anonymization.

In the original dataset, approximately 51.43% of the records are unique based on their combination of quasi-identifying attributes. This explains why the highest prosecutor risk is 100%.

After applying k-anonymity with $k = 5$, every non-suppressed record belongs to an equivalence class containing at least five records. Therefore, the maximum theoretical prosecutor risk is:

$$
Risk_{max} = \frac{1}{k} = \frac{1}{5} = 0.2 = 20\%
$$

This corresponds to the result obtained in ARX, where the highest prosecutor risk decreases from 100% to 20%.

The average prosecutor risk also decreases from approximately 64.66% to 2.18%. Furthermore, the percentage of records affected by the highest risk decreases from 51.43% to approximately 1.14%.

The percentage of unique records in the sample decreases from 51.43% to 0%. This means that no non-suppressed record can be uniquely distinguished based on its combination of quasi-identifying attributes.

By comparing the risk distribution plots, we can also observe that a large proportion of the original records have a high re-identification risk. After anonymization, most records have a much lower risk, and none of the analyzed records exceed the 20% risk threshold.

The estimated percentage of population uniques also decreases to 0%. However, these values are based on different statistical population models (PITMAN for the original data and DANKAR for the transformed data), so they should be interpreted as estimates.

Overall, applying k-anonymity with $k = 5$ significantly reduces the re-identification risk through generalization and suppression. However, the risk is not completely eliminated, since some records still have a re-identification risk of 20%.

**d. Privacy model with lower re-identification risk and the same or higher classification accuracy**

To find a privacy model that provides a lower re-identification risk while maintaining or improving the classification accuracy, we tested different values of $k$ in the k-anonymity model.

We kept the same anonymization settings and classification parameters as in the previous exercises. We selected `sex` as the target variable and used the remaining eight attributes as input features. The classification was performed using 10-fold cross-validation.

We tested k-anonymity with $k = 6$, $k = 7$ and $k = 10$ and compared the results with the ones obtained for $k = 5$.

| Measure | k = 5 | k = 6 | k = 7 | k = 10 |
|---|---:|---:|---:|---:|

| Random Forest accuracy | 71.34474% | 69.66050% | 72.45209% | 71.09940% |

We can observe that increasing $k$ reduces the maximum possible prosecutor risk, but does not necessarily improve the classification accuracy.

For $k = 6$, the accuracy of Logistic Regression slightly increases, while the accuracy of Random Forest decreases.

For $k = 7$, both Logistic Regression and Random Forest achieve a higher accuracy compared to $k = 5$, while the accuracy of Naive Bayes remains almost unchanged.

For $k = 10$, the maximum theoretical prosecutor risk is reduced to 10%, but the accuracy decreases for all three classification algorithms compared to $k = 5$.

Based on these results, we selected k-anonymity with $k = 7$.

With this privacy model, each non-suppressed record belongs to an equivalence class containing at least seven records. Therefore, the maximum theoretical prosecutor risk is:

$$
Risk_{max} \leq \frac{1}{7} \approx 14.29\%
$$

Compared to k-anonymity with $k = 5$, the maximum possible prosecutor risk decreases by:

$$
20 - \frac{100}{7} \approx 5.71
$$

This corresponds to a decrease of approximately 5.71 percentage points.

The improvement in classification accuracy for Logistic Regression is:

$$
74.25900 - 74.19601 = 0.06299
$$

The improvement in classification accuracy for Random Forest is:

$$
72.45209 - 71.34474 = 1.10735
$$

The applied generalization levels remain the same for all four configurations:

`[0, 3, 0, 0, 2, 1, 1, 1, 0]`

The differences in classification accuracy may be related to changes in record suppression and variation in the classification evaluation. Therefore, the observed improvements do not necessarily mean that increasing $k$ generally leads to better classification accuracy.

Overall, k-anonymity with $k = 7$ provides the best privacy-utility trade-off among the tested configurations that satisfy the requirements. It reduces the maximum prosecutor risk while achieving a higher measured classification accuracy for Logistic Regression and Random Forest compared to k-anonymity with $k = 5$.

## **Problem 3**

The idea of the blocking approach by Karakasidis and Verykios is to use a shared reference table (RT) to create initial clusters. Each data owner then assigns their vectors to these clusters using the nearest neighbor criterion based on the Dice coefficient.

### Step 1: Define the vectors

The reference table contains the following vectors:

$$
r_1 = (1,1,0,0,1), \quad r_2 = (1,0,0,0,1)
$$

$$
r_3 = (0,0,1,1,0), \quad r_4 = (0,1,1,1,0)
$$

The vectors of $DS_1$ are:

$$
x_1 = (0,1,0,0,1), \quad x_2 = (0,0,1,1,1)
$$

The vectors of $DS_2$ are:

$$
y_1 = (1,0,0,0,1), \quad y_2 = (1,0,1,1,0)
$$

### Step 2: Create the initial clusters

To create the two initial clusters, we calculate the similarity between the vectors of the reference table using the Dice coefficient:

$$
Dice(A,B) = \frac{2h}{a+b}
$$

where $a$ and $b$ are the numbers of 1-bits in the two vectors, and $h$ is the number of 1-bits they have in common.

$$
Dice(r_1,r_2) = \frac{2 \cdot 2}{3+2} = 0.8
$$

$$
Dice(r_1,r_3) = \frac{2 \cdot 0}{3+2} = 0
$$

$$
Dice(r_1,r_4) = \frac{2 \cdot 1}{3+3} \approx 0.333
$$

$$
Dice(r_2,r_3) = \frac{2 \cdot 0}{2+2} = 0
$$

$$
Dice(r_2,r_4) = \frac{2 \cdot 0}{2+3} = 0
$$

$$
Dice(r_3,r_4) = \frac{2 \cdot 2}{2+3} = 0.8
$$

The similarity matrix is:

| | $r_1$ | $r_2$ | $r_3$ | $r_4$ |
|---|---:|---:|---:|---:|
| $r_1$ | 1 | 0.8 | 0 | 0.333 |
| $r_2$ | 0.8 | 1 | 0 | 0 |
| $r_3$ | 0 | 0 | 1 | 0.8 |
| $r_4$ | 0.333 | 0 | 0.8 | 1 |

Since $r_1$ and $r_2$ have a high similarity, as well as $r_3$ and $r_4$, we form the following two clusters:

$$
C_1 = \{r_1,r_2\}
$$

$$
C_2 = \{r_3,r_4\}
$$

### Step 3: Assign the vectors from $DS_1$

Each vector is compared with the vectors from the reference table and assigned to the cluster containing its most similar reference vector.

For $x_1$:

$$
Dice(x_1,r_1) = \frac{2 \cdot 2}{2+3} = 0.8
$$

$$
Dice(x_1,r_2) = \frac{2 \cdot 1}{2+2} = 0.5
$$

$$
Dice(x_1,r_3) = \frac{2 \cdot 0}{2+2} = 0
$$

$$
Dice(x_1,r_4) = \frac{2 \cdot 1}{2+3} = 0.4
$$

The highest similarity is with $r_1$, so $x_1$ is assigned to $C_1$.

For $x_2$:

$$
Dice(x_2,r_1) = \frac{2 \cdot 1}{3+3} \approx 0.333
$$

$$
Dice(x_2,r_2) = \frac{2 \cdot 1}{3+2} = 0.4
$$

$$
Dice(x_2,r_3) = \frac{2 \cdot 2}{3+2} = 0.8
$$

$$
Dice(x_2,r_4) = \frac{2 \cdot 2}{3+3} \approx 0.667
$$

The highest similarity is with $r_3$, so $x_2$ is assigned to $C_2$.

### Step 4: Assign the vectors from $DS_2$

We repeat the same procedure for the second dataset.

For $y_1$:

$$
Dice(y_1,r_1) = \frac{2 \cdot 2}{2+3} = 0.8
$$

$$
Dice(y_1,r_2) = \frac{2 \cdot 2}{2+2} = 1
$$

$$
Dice(y_1,r_3) = \frac{2 \cdot 0}{2+2} = 0
$$

$$
Dice(y_1,r_4) = \frac{2 \cdot 0}{2+3} = 0
$$

The highest similarity is with $r_2$, so $y_1$ is assigned to $C_1$.

For $y_2$:

$$
Dice(y_2,r_1) = \frac{2 \cdot 1}{3+3} \approx 0.333
$$

$$
Dice(y_2,r_2) = \frac{2 \cdot 1}{3+2} = 0.4
$$

$$
Dice(y_2,r_3) = \frac{2 \cdot 2}{3+2} = 0.8
$$

$$
Dice(y_2,r_4) = \frac{2 \cdot 2}{3+3} \approx 0.667
$$

The highest similarity is with $r_3$, so $y_2$ is assigned to $C_2$.

### Step 5: Comparison by the third party

After assigning all the vectors, we obtain the following results:

| Cluster | Reference table | $DS_1$ | $DS_2$ |
|---|---|---|---|
| $C_1$ | $r_1, r_2$ | $x_1$ | $y_1$ |
| $C_2$ | $r_3, r_4$ | $x_2$ | $y_2$ |

Both data owners send their data together with the corresponding cluster indices to the third party. The third party only compares vectors that belong to the same cluster.

Therefore, the following vectors will be compared:

$$
\begin{pmatrix}0\\1\\0\\0\\1\end{pmatrix}
\longleftrightarrow
\begin{pmatrix}1\\0\\0\\0\\1\end{pmatrix}
$$

$$
\begin{pmatrix}0\\0\\1\\1\\1\end{pmatrix}
\longleftrightarrow
\begin{pmatrix}1\\0\\1\\1\\0\end{pmatrix}
$$

The pairs $(x_1,y_2)$ and $(x_2,y_1)$ will not be compared because they belong to different clusters.

Without blocking, we would have to perform $2 \cdot 2 = 4$ comparisons. By using this approach, we only have to perform 2 comparisons, which reduces the number of comparisons by 50%.
