## **Problem 1**

a. The three main phases are development, implementation, and evaluation. First, the CDSS is designed and developed. Then, it is introduced into clinical practice. Finally, it is evaluated to see if it works correctly and actually helps healthcare professionals.

b. A learning healthcare system is a system that continuously uses data from everyday patient care to improve future healthcare. For example, information about patients and their treatments can be analyzed to find out what works best. These findings are then used to improve medical decisions and treatments. This creates a cycle where healthcare constantly learns from previous experience.

c. Alert-fatigue happens when healthcare professionals receive too many warnings or notifications from a system. After some time, they might start ignoring them, even when some alerts are actually important.

d. In PCA (Principal Component Analysis), we try to maximize the variance of the data along the new principal components. The first principal component captures as much variance as possible, and the following components capture the most remaining variance while being orthogonal to the previous ones. This helps us reduce the number of dimensions while keeping as much information as possible.

e. Differential privacy is a way to protect people's personal information when analyzing or sharing data. The idea is that the results should not change too much depending on whether one specific person's data is included or not. This is usually achieved by adding carefully calculated random noise to the results. The amount of noise depends on the privacy parameter ε (epsilon) and the sensitivity of the calculation. A smaller epsilon means stronger privacy protection.

f. K-anonymity means that every person in a dataset should be indistinguishable from at least k−1 other people based on certain attributes that could help identify them, such as age or gender. For example, with k = 5, every combination of these identifying attributes must appear at least 5 times. This can be achieved by generalizing information, like replacing an exact age with an age range, or by removing some details.

### **Problem 2**

a. To find the largest k, we need to count how many people have the same combination of the two quasi-identifiers (Gender and Age-range).

| Gender | Age-range | Number of people |
|---|---|---|
| Female | 19–24 | 3 |
| Female | 25–30 | 4 |
| Male | 19–24 | 4 |
| Male | 25–30 | 3 |

The smallest group contains 3 people. Since k-anonymity requires every combination of quasi-identifiers to appear at least k times, the largest possible value is:

**Answer: k = 3.**

b. Since we want to compare the two age groups, we need to keep the Age-range information. Instead, we can remove the Gender information by replacing it with `*` for every person.

This gives us two groups:

| Gender | Age-range | Number of people | HIV positive | HIV negative |
|---|---|---|---|---|
| * | 19–24 | 7 | 4 | 3 |
| * | 25–30 | 7 | 1 | 6 |

Now, each combination of quasi-identifiers appears 7 times, so the dataset satisfies **k = 7**.

We can still compare the HIV-positive rates:

- Age 19–24: 4/7 ≈ 57.1%
- Age 25–30: 1/7 ≈ 14.3%

Therefore, based on this dataset, people aged 19–24 are more likely to have an HIV-positive status than people aged 25–30.

c. This time, we want to compare men and women, so we need to keep the Gender information. Instead, we can generalize the Age-range by replacing both age groups with one larger range, **19–30**.

The resulting groups are:

| Gender | Age-range | Number of people | HIV positive | HIV negative |
|---|---|---|---|---|
| Female | 19–30 | 7 | 2 | 5 |
| Male | 19–30 | 7 | 3 | 4 |

Both groups now contain 7 people, so the dataset satisfies **k = 7**.

We can still compare the HIV-positive rates:

- Female: 2/7 ≈ 28.6%
- Male: 3/7 ≈ 42.9%

Therefore, based on this dataset, men are more likely to have an HIV-positive status than women.

**!!!Important!!!** In both b. and c. we only generalize the quasi-identifier that is not needed for the comparison. This allows us to achieve k = 7 while still keeping the information relevant to the research question. Also, when publishing the anonymized dataset, we should remove the ID column, since keeping a unique identifier for every person could make re-identification possible.
